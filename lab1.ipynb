```python
# Create the initial urn

urn = ["red"] * 3 + ["blue"] * 4 + ["black"] * 2


#Roll a six-sided die

die_roll = np.random.randint(1, 7)

#Add a ball according to the die result

if die_roll in (2, 3, 5):
    urn.append("black")
elif die_roll == 6:
    urn.append("red")
else:
    urn.append("blue")

#Draw one ball randomly

drawn_ball = np.random.choice(urn)

#Display the results

print("Die roll:", die_roll)
print("Drawn ball:", drawn_ball)
print("Number of balls in the urn:", len(urn))#
```
