## Part A: Conditional Independence in Bayesian Networks

The Bayesian network contains four variables:

- **S**: Whether the email is spam.
- **O**: Whether the email contains the word "oferta".
- **L**: Whether the email contains links.
- **M**: Whether the email is long.

The network structure is defined by the following directed edges:

- S → O
- S → L
- S → M
- L → M

### Conditional Independence

In this Bayesian network, O and L are conditionally independent given S:

**P(O, L | S) = P(O | S) × P(L | S)**

Furthermore, O and M are conditionally independent given S:

**P(O, M | S) = P(O | S) × P(M | S)**

However, O and L are not necessarily independent without knowing S, since both variables depend on whether the email is spam.

Finally, M depends on both S and L, as specified by the network structure.

```python
# Prior probabilities
p_spam = 0.4
p_not_spam = 0.6

# Probability of containing "oferta"
p_offer_given_spam = 0.7
p_offer_given_not_spam = 0.1

# Probability of containing links
p_links_given_spam = 0.8
p_links_given_not_spam = 0.3

# Probability of being a long email
p_long_given_spam_and_links = 0.9
p_long_given_spam_and_no_links = 0.5
p_long_given_not_spam_and_links = 0.6
p_long_given_not_spam_and_no_links = 0.2
```

```python
def calculate_spam_probability(O, L, M):
    # Calculate the joint probability when the email is spam
    p_O_spam = (
        p_offer_given_spam if O == 1
        else 1 - p_offer_given_spam
    )

    p_L_spam = (
        p_links_given_spam if L == 1
        else 1 - p_links_given_spam
    )

    p_M_spam = (
        p_long_given_spam_and_links if L == 1
        else p_long_given_spam_and_no_links
    )

    if M == 0:
        p_M_spam = 1 - p_M_spam

    joint_spam = (
        p_spam * p_O_spam * p_L_spam * p_M_spam
    )

    # Calculate the joint probability when the email is not spam
    p_O_not_spam = (
        p_offer_given_not_spam if O == 1
        else 1 - p_offer_given_not_spam
    )

    p_L_not_spam = (
        p_links_given_not_spam if L == 1
        else 1 - p_links_given_not_spam
    )

    p_M_not_spam = (
        p_long_given_not_spam_and_links if L == 1
        else p_long_given_not_spam_and_no_links
    )

    if M == 0:
        p_M_not_spam = 1 - p_M_not_spam

    joint_not_spam = (
        p_not_spam
        * p_O_not_spam
        * p_L_not_spam
        * p_M_not_spam
    )

    # Apply Bayes' theorem
    probability = joint_spam / (joint_spam + joint_not_spam)

    return probability
```

```python
probability = calculate_spam_probability(O=1, L=1, M=1)

print(f"Probability of spam: {probability:.4f}")
print(f"Percentage: {probability * 100:.2f}%")
```

```python
# Test all possible combinations of email characteristics
print("Email Classification Results")
print("-" * 65)
print(f"{'O':<5}{'L':<5}{'M':<5}{'P(Spam)':<15}{'Classification'}")
print("-" * 65)

for O in [0, 1]:
    for L in [0, 1]:
        for M in [0, 1]:
            probability = calculate_spam_probability(O, L, M)

            if probability > 0.5:
                classification = "Spam"
            else:
                classification = "Not Spam"

            print(
                f"{O:<5}{L:<5}{M:<5}"
                f"{probability:<15.4f}{classification}"
            )
```

### Conclusion

The Bayesian network classifies emails by calculating the posterior probability that an email is spam, given its characteristics: whether it contains the word "oferta", whether it contains links, and whether it is long.

The model evaluates all eight possible combinations of these three binary variables. An email is classified as spam if its posterior probability exceeds 0.5; otherwise, it is classified as not spam.

For example, an email containing "oferta", containing links, and being long has a spam probability of approximately 94.92%. Therefore, the model classifies this email as spam.
