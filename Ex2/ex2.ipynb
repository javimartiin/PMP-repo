```python
import numpy as np

print("PMP - Laboratory 2")
print("Bayes' Theorem")
```

```python
# Given probabilities
p_disease = 0.01
sensitivity = 0.95
specificity = 0.90

# Complementary probabilities
p_no_disease = 1 - p_disease
false_positive_rate = 1 - specificity

print("Probability of disease:", p_disease)
print("Sensitivity:", sensitivity)
print("Specificity:", specificity)
print("False positive rate:", false_positive_rate)
```

```python
# Calculate the probability of a positive test
p_positive = (
    sensitivity * p_disease
    + false_positive_rate * p_no_disease
)

# Apply Bayes' theorem
p_disease_given_positive = (
    sensitivity * p_disease
    / p_positive
)

print("Probability of a positive test:", p_positive)
print("Probability of disease given a positive test:",
      p_disease_given_positive)
print("Percentage:",
      p_disease_given_positive * 100, "%")
```

```python
# Calculate the probability of a positive test
p_positive = (
    sensitivity * p_disease
    + false_positive_rate * p_no_disease
)

# Apply Bayes' theorem
p_disease_given_positive = (
    sensitivity * p_disease / p_positive
)

print("Probability of a positive test:", p_positive)
print("Probability of disease given a positive test:",
      p_disease_given_positive)
print("Percentage:", p_disease_given_positive * 100, "%")
```

```python
# Number of people to simulate
n_people = 1_000_000

# Simulate whether each person has the disease
has_disease = np.random.random(n_people) < p_disease

# Simulate test results
test_positive = np.zeros(n_people, dtype=bool)

# Positive results among people with the disease
test_positive[has_disease] = (
    np.random.random(np.sum(has_disease)) < sensitivity
)

# Positive results among people without the disease
test_positive[~has_disease] = (
    np.random.random(np.sum(~has_disease)) < false_positive_rate
)

# Calculate the simulated probability
n_positive = np.sum(test_positive)
n_disease_and_positive = np.sum(has_disease & test_positive)

p_simulated = n_disease_and_positive / n_positive

print("Number of people simulated:", n_people)
print("Number of positive tests:", n_positive)
print("Number of sick people with a positive test:",
      n_disease_and_positive)
print("Simulated probability:", p_simulated)
print("Simulated percentage:", p_simulated * 100, "%")
```

```python
# Compare theoretical and simulated probabilities
absolute_error = abs(p_disease_given_positive - p_simulated)

print("Theoretical probability:", p_disease_given_positive)
print("Simulated probability:", p_simulated)

print("Theoretical percentage:",
      p_disease_given_positive * 100, "%")

print("Simulated percentage:",
      p_simulated * 100, "%")

print("Absolute error:", absolute_error)
print("Absolute error in percentage points:",
      absolute_error * 100)
```

```python
# Desired probability of disease given a positive test
target_probability = 0.50

# Calculate the required specificity
required_specificity = (
    1 - (sensitivity * p_disease)
    / (p_no_disease * target_probability / (1 - target_probability))
)

print("Required specificity:", required_specificity)
print("Required specificity percentage:",
      required_specificity * 100, "%")
```

# Laboratory 2: Bayes' Theorem

## Introduction

The aim of this exercise is to apply Bayes' theorem to calculate the probability that a person has a disease given a positive diagnostic test result.

The disease affects 1% of the population. The diagnostic test has a sensitivity of 95% and a specificity of 90%.

The exercise consists of three parts:
- **Part A:** Calculate the theoretical probability of having the disease given a positive test result.
- **Part B:** Simulate a large population in Python and compare the estimated probability with the theoretical result.
- **Part C:** Determine the minimum specificity required for the probability of having the disease after a positive test to reach 50%.

## Part A: Theoretical Calculation

Bayes' theorem allows us to calculate the probability of having a disease given a positive test result:

\[
P(B\mid +)=\frac{P(+\mid B)P(B)}{P(+)}
\]

The probability of a positive test result is:

\[
P(+)=P(+\mid B)P(B)+P(+\mid \neg B)P(\neg B)
\]

Given the sensitivity of 95% and the specificity of 90%, the false positive rate is:

\[
P(+\mid \neg B)=1-0.90=0.10
\]

Substituting the values into Bayes' theorem:

\[
P(B\mid +)=
\frac{0.95\cdot0.01}
{0.95\cdot0.01+0.10\cdot0.99}
\]

Therefore,

\[
\boxed{P(B\mid +)\approx0.08756=8.76\%}
\]

Although the test has high sensitivity, the disease is rare. Consequently, false positive results among the much larger healthy population substantially affect the probability that a person is actually ill after receiving a positive result.

## Part B: Python Simulation

To verify the theoretical result, a simulation of 1,000,000 people is performed using Python and NumPy.

First, each person is randomly classified as either having the disease or being healthy, according to the disease prevalence of 1%. Then, a diagnostic test result is simulated using a sensitivity of 95% and a specificity of 90%.

Finally, the probability of having the disease given a positive test result is estimated by dividing the number of people who are both sick and test positive by the total number of positive test results.

The theoretical probability obtained in Part A is approximately 8.76%, whereas the simulated probability is approximately 8.73%. The absolute difference is about 0.0292 percentage points.

The simulated result is very close to the theoretical value. The small difference is due to random variation in the simulation. Increasing the number of simulated people generally improves the stability of the estimate.

## Part C: Required Specificity

The objective is to determine the specificity required for the probability of having the disease given a positive test result to reach 50%.

Using Bayes' theorem:

\[
P(B\mid +)=
\frac{P(+\mid B)P(B)}
{P(+\mid B)P(B)+P(+\mid\neg B)P(\neg B)}
\]

Let \(s\) denote the specificity. Since the false positive rate is \(1-s\), we obtain:

\[
0.50=
\frac{0.95\cdot0.01}
{0.95\cdot0.01+(1-s)\cdot0.99}
\]

Rearranging the equation gives:

\[
0.95\cdot0.01=(1-s)\cdot0.99
\]

Therefore:

\[
s=1-\frac{0.95\cdot0.01}{0.99}
\]

The required specificity is:

\[
\boxed{s\approx0.990404=99.04\%}
\]

Thus, the test must have a specificity of approximately 99.04% for the probability of having the disease after a positive result to reach 50%, given the stated prevalence and sensitivity.
